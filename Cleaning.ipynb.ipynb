```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic_Unclean_Large.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

```python
df['Fare'].fillna(df['Fare'].median(), inplace=True)
```

```python
df['Sex'].fillna(df['Sex'].mode()[0], inplace=True)
```

```python
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.info()
```

```python
df['Age'] = df['Age'].astype(int)
df['Fare'] = df['Fare'].astype(float)
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 12, 18, 60, 100],
    labels=['Child', 'Teen', 'Adult', 'Senior']
)
```

```python
df['Fare_Category'] = pd.cut(
    df['Fare'],
    bins=[0, 50, 150, 300, 600],
    labels=['Low', 'Medium', 'High', 'Very High']
)
```

```python
df['Survival_Status'] = df['Survived'].map({
    0: 'Not Survived',
    1: 'Survived'
})
```

```python
df.to_csv("Titanic_Cleaned_Data.csv", index=False)
```

### Data Cleaning Summary
- Loaded Titanic dataset into Google Colab
- Identified missing values using isnull().sum()
- Filled numeric columns (Age, Fare) using median
- Filled categorical columns (Sex, Embarked) using mode
- Removed duplicate records
- Converted data types for accuracy
- Created new columns such as Age Group and Fare Category
- Saved cleaned dataset for further analysis
